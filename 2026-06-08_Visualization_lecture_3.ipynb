```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# set a random seed during data generation to ensure the same data are generated eat time the code is run
np.random.seed(613)

# x is integers from 0 to 49
x = np.arange(50)

y1 = np.random.randint(0, 100, 50)
# if you want y2 to generate hte same data as in y1, you need to reset the random seed
#np.random.seed(613)
y2 = np.random.randint(0, 100, 50)
```

```python
# define a custom font dictionary to used in plots

font1 = {"family" : "serif", 
    "color" : "indigo",
    "size" : 20}
```

```python
fig, ax = plt.subplots(1,1, figsize=(4,3))

ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# bbox_to_anchor relative to axis plot area. 0,0 is bottom left, 1,1 is top right, 1, 0 is bottom right
ax.legend(frameon=True, ncol=2, shadow=True, loc="upper left", bbox_to_anchor=(1, 0.5))

# text based on axis values
ax.text(10, 55, "This is sample text!", ha="left")

# add an arrow. xy is the data point to point at, xytext is where to put the text
# create a small dictionary to describe the arrow 
ax.annotate("This is important!", xy=(10, 92), xytext=(20, 95), 
    arrowprops=dict(facecolor="black"), color="red")

ax.annotate("This is also important!", xy=(10, 22), xytext=(20, 25), 
    arrowprops=dict(arrowstyle="wedge",
    color="hotpink"), 
    color="red")

# remove y axis labels and ticks
#ax.yaxis.set_major_locator(plt.NullLocator())

# remove x axis labels but keep tick marks
#ax.xaxis.set_major_formatter(plt.NullFormatter())

# adjust the total number of major ticks on the x axis
ax.xaxis.set_major_locator(plt.MaxNLocator(3))

# specify an interval to set up major tick marks
ax.xaxis.set_major_locator(plt.MultipleLocator(5))

# set x axis label using custom font dictionary
ax.set_xlabel("x axis label", fontdict=font1)

# rotate axis labels
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
fig, ax = plt.subplots()

# create axis area, 20 inch square
ax.axis([0, 20, 0, 20])

# Various ways to position custom text within the plot

# use transform=ax.transData to connect text to a specific data point with x, y. values
# I believe this is the default, without transform, text annotations appear beside the data points
# based on the axes
ax.text(1, 5, ". ax.transData:(1, 5)", transform=ax.transData)

# can also position text based on relative fraction of the axis value.
ax.text(0.5, 0.1, ". ax.transAxes:(0.5, 0.1)", transform=ax.transAxes)

# fraction relevant to the whole canvas
ax.text(0.5, 0.5, ". fig.transFigure:(0.5, 0.5)", transform=fig.transFigure)
```
