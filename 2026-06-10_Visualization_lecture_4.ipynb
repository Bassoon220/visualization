```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python

x1 = np.arange(50)
np.random.seed(613)
y1 = np.random.randint(0, 75, 50)

x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
plt.style.use("ggplot")

font1 = {"family" : "serif", 
    "color" : "indigo",
    "size" : 10}

font2 = {"family" : "monospace", 
    "color" : "indigo",
    "size" : 10}

fig, axs = plt.subplots(1, 2, figsize=(7,3))

axs[0].scatter(x1, y1)
axs[1].bar(x2, y2)

axs[0].set_xlabel("x axis label", fontdict=font1)

fig.show()
```

```python
# trying tight layout

# need to reset style to default
plt.style.use("default")

# set up subplots in a mosaic which doesn't strictly follow a grid but also kinda does
#fig, someaxes = plt.subplot_mosaic([["ax1", "ax3"], ["ax2", "ax3"]], figsize=(7,4), label="constrained")
fig, someaxes = plt.subplot_mosaic([["ax1", "ax3"], ["ax2", "ax3"]], figsize=(7,4))


someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel("ax1 x axis label", fontsize=10, fontfamily="monospace")

plt.tight_layout()

fig.show()
```

```python
from PIL import Image
import requests
from io import BytesIO

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

# sometimes if you don't specify headers then wikipedia won't let you download images
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

# extract the image from the specified url
response=requests.get(url,headers=headers)

# download the contents of the response
image_file=BytesIO(response.content)

# open the image, then show it
image=Image.open(image_file)
image
```

```python
# create two plots on the same set of axes
# also put an image on the figure

x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7,3))

ax.bar(x, y1, color="indigo")

ax.plot(x, y2, color="red")

# create a box for the image
ax_image_box = fig.add_axes([0.15, # percentage of x axis relative to entire figure, where to put figure
    0.5, # where to place figure, percentage of y axis
    0.15, # image width
    0.35]) # image height

# add the image to the box, use imshow, then hide the axes of the image plot
ax_image_box.imshow(image)
ax_image_box.axis("off")
# in the example, the luffy image is narrow, so it is shown in the middle of the plot area
# to adjust position, you should change the first two numbers when creating the box, the position
# but technically you could also adjust the image position through the width/height too, it's prob weird

fig.show()
```
