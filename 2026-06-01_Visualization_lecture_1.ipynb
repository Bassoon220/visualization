```python
# generate some dummy data
np.random.seed(613)

# generate x data, integers from 0-49
x = np.arange(50)

# generate y data, random integers from 0-99, make 50 numbers
y = np.random.randint(0, 100, 50)
```

```python
y.max()
```

```python
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x, y);
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x, y);
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y);
```

```python
fig, ax = plt.subplots(figsize=(5,3))

ax.hist(y, bins=10)
```

```python
fig, ax = plt.subplots(figsize=(5,3))

ax.hist(y, bins=10)

ax.set_title("plot title")
ax.set_ylabel("y axis label")
ax.set_xlabel("x axis label")
plt.tight_layout()
```

```python
# create font libraries/dictionaries

font1 = {"family" : "arial", 
    "color" : "blue",
    "size" : 20}

font2 = {"family" : "monospace", 
    "color" : "green",
    "size" : 14}
```

```python
fig, ax = plt.subplots(figsize=(5,3))

ax.scatter(x, y)

# use custom font dictionaries in the plot
ax.set_title("plot title", fontdict=font1)
ax.set_ylabel("y axis label", fontdict=font2)
ax.set_xlabel("x axis label", fontdict=font2)
plt.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5,3))


ax.plot(x, y, marker="", color="lightblue", linestyle="--", linewidth=2)
ax.scatter(x, y, marker="*", color="indigo", zorder=2)


ax.set_title("plot title", fontdict=font1)
ax.set_ylabel("y axis label", fontdict=font2)
ax.set_xlabel("x axis label", fontdict=font2)
plt.tight_layout()
#fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5,3))


#ax.plot(x, y, marker="", color="lightblue", linestyle="--", linewidth=2)

# this is a better way, maybe, to get separate plot markers and connecting lines 
ax.plot(x, y, marker="*", color="indigo", zorder=2,
    linestyle="--", linewidth=2,
    markeredgecolor="orange",
    markerfacecolor="teal")

ax.grid(axis="y", color="dodgerblue", alpha=0.5, linewidth=1, linestyle="--")

ax.set_title("plot title", fontdict=font1)
ax.set_ylabel("y axis label", fontdict=font2)
ax.set_xlabel("x axis label", fontdict=font2)
plt.tight_layout()
#fig.show()
```
